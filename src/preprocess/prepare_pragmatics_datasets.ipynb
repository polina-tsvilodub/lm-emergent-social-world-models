```python
def get_incorrect_option(row, correct_col="emotion_label", options_col="emotion_choices"):
    correct = row[correct_col]
    options = row[options_col]
    incorrect_option = list(filter(lambda x: x!= correct, options))
    if correct_col == "cause_label":
        incorrect_option = np.random.choice(incorrect_option, size=3, replace=False)
        return incorrect_option
    return incorrect_option
```

```python
def create_choices_prompt(row, correct, incorrect_options):
    # TODO: create options in randomized order, make sure to record which one is the correct one (with the numbered prefix)
    options = [row[correct]] + [row[opt] for opt in incorrect_options]
    np.random.shuffle(options)
    correct_index = options.index(row[correct]) + 1
    options_prompt = "Choices:\n" + "\n".join([f"{i+1}. {opt}" for i, opt in enumerate(options)])
    # get the incorrect options only
    incorrect_prompt = options_prompt.replace(f"{correct_index}. {row[correct]}", "").replace("Choices:\n", "").strip()
    incorrect_options_processed = [o for o in incorrect_prompt.split("\n") if len(o.strip()) > 0]
    return options_prompt, f"{correct_index}. {row[correct]}", incorrect_options_processed
```

```python
emo_eu_en_emotion[[
    "emotion_choices_prompt",
    "emotion_correct",
    "incorrect_emotions_processed"
]] = pd.DataFrame(emo_eu_en_emotion.apply(create_choices_prompt, axis=1, correct="emotion_label", incorrect_options=["incorrect_emotion_1", "incorrect_emotion_2", "incorrect_emotion_3", "incorrect_emotion_4", "incorrect_emotion_5"]).to_list(), index=emo_eu_en_emotion.index)
```

```python
# construct an MC version of the entity tracking dataset by Kim and Schuster
import json
import csv
import random
import re
from pathlib import Path

INPUT_PATH = "../../../boxes-dataset-v1/few_shot_boxes_nso_exp2_max3/test-subsample-states-t5.jsonl"
OUTPUT_PATH = "../../behavioral_eval_stimuli/log_p_curated/multiple_choice_entity_tracking.csv"
NUM_INCORRECT = 5  # number of distractor options per example
MASK_TOKEN = "<extra_id_0>"  # assumed masking prefix

PROMPT_TEMPLATE = """Given the description after "Description:", write a true statement about all boxes and their contents according to the description after "Statement:". Format the statement as Box 0 contains the A, Box 1 contains the B, Box 2 contains the C, Box 3 contains the D, Box 4 contains the E, Box 5 contains the F, Box 6 contains the G. A, B, C, D, E, F, G are placeholders for the contents of each box.

Description: {desc}"""
ANSWER_TEMPLATE = "\nStatement: {answer}"

def extract_description(full_input):
    """
    Extract description text before the masking token.
    """
    if MASK_TOKEN in full_input:
        desc = full_input.split(MASK_TOKEN)[0].strip()
        answer = full_input.split(".")[-2].strip().replace(MASK_TOKEN, "").strip()
    else:
        desc = full_input.strip()
        answer = full_input.split(".")[-2].strip()
    return desc, answer


def extract_objects_from_text(text):
    """
    Extract candidate object nouns from description using simple regex.
    Assumes objects appear as 'the X' or standalone nouns.
    """
    # Capture words after 'the'
    candidates = re.findall(r"\bthe ([a-zA-Z\-]+)", text)
    print("--- candidates ---- ", candidates)
    # Remove duplicates
    return list(set(candidates))


def parse_correct_statement(statement):
    """
    Parse correct answer into a list of box contents.
    Returns list of strings per box.
    """
    boxes = re.findall(r"Box \d+ contains ([^,]+(?:,?\.[^B]*)?)", statement)
    boxes = [b.split(".")[0] for b in boxes]
    print("boxes : ----", boxes)
    return [b.strip() for b in boxes]


def generate_incorrect_options(objects, num_boxes, true_statement):
    """
    Generate incorrect full statements by randomly assigning
    objects across boxes.
    """
    incorrect_options = set()

    # Ensure enough objects for sampling
    if len(objects) == 0:
        return []

    while len(incorrect_options) < NUM_INCORRECT:
        shuffled = objects[:]
        random.shuffle(shuffled)

        # Random partition of objects across boxes
        partitions = []
        idx = 0
        for _ in range(num_boxes):
            # Random 0-2 objects per box
            k = random.randint(0, min(2, len(shuffled) - idx))
            contents = shuffled[idx:idx + k]
            idx += k
            if contents:
                partitions.append(" and ".join(f"the {o}" for o in contents))
            else:
                partitions.append("nothing")

        candidate = ", ".join(
            f"{partitions[i]}"
            for i in range(num_boxes)
        )

        if candidate != true_statement:
            incorrect_options.add(candidate)

    return list(incorrect_options)


def main():
    rows = []
    num_samples = 0
    with open(INPUT_PATH, "r") as f:
        # take the first 670 examples 
        for line in f:
            sample = json.loads(line)
            if num_samples >= 670:
                break
            num_samples += 1

            full_input = sample["sentence_masked"]
            print("full input", full_input)
            correct = sample["masked_content"].replace(MASK_TOKEN, "").strip() + "."
            print("correct ", correct)
            desc, answer = extract_description(full_input)
            target_box_number = re.findall(r"\d+", answer)
            print("-- answer and extracted target box number: ---- ", answer, target_box_number)
            context = PROMPT_TEMPLATE.format(desc=desc)
            answer_prefix = ANSWER_TEMPLATE.format(answer=answer)
            objects = extract_objects_from_text(desc)
            print("-- extracted objects: ", objects)
            box_contents = parse_correct_statement(desc)
            print("box contents : ----", box_contents)
            num_boxes = len(box_contents)
            print("num boxes : ----", num_boxes)
            # use the contents of non-target boxes as incorrect options
            incorrect = [f"{content}." for i, content in enumerate(box_contents) if i != target_box_number]
            incorrect_filtered = [i for i in incorrect if i != correct]
            incorrect_samples = np.random.choice(incorrect_filtered, size=min(num_boxes-1, len(incorrect_filtered)), replace=False).tolist()
            print(len(incorrect_samples), "incorrect options generated from non-target boxes: ", incorrect_samples)
            # incorrect = generate_incorrect_options(
            #     objects, num_boxes, correct
            # )

            rows.append({
                "context": context + answer_prefix,
                "correct": correct,
                "incorrect": incorrect_samples
            })

    with open(OUTPUT_PATH, "w", newline="") as csvfile:
        writer = csv.DictWriter(
            csvfile,
            fieldnames=["context", "correct", "incorrect"]
        )
        writer.writeheader()
        writer.writerows(rows)

    print(f"Saved to {OUTPUT_PATH}")


if __name__ == "__main__":
    main()
```
